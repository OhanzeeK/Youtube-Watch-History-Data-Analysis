```python
import pandas as pd
import numpy as np
import matplotlib as plt
import json
```

```python
#Opening watch history JSON
with open(r"C:\Users\0106o\OneDrive\Pictures\Documents\Datasets\Youtube Project Files\history\watch-history.json", encoding="utf-8") as f:
    data = json.load(f)

df = pd.json_normalize(data)
df
```

```python
#Checking the dimensions of the dataset
df.shape
```

```python
#Checking the types and names of each column
df.info()
```

```python
#Checking which columns have the most nulls to determine which are worth keeping and which are being dropped
df.isnull().sum()
```

```python
#Dropping duplicates and checking if counts changed

df = df.drop_duplicates(subset = ['title', 'time'])
df.count()
```

```python
#Creating a new source column to drop all videos that were forced ads from youtube
df["source"] = df["details"].apply(lambda x: x[0]["name"] if isinstance(x, list) and len(x) > 0 else None)
df
```

```python
#Filtering out ads from dataset
df = df[df['source'] != 'From Google Ads']
df
```

```python
# df = df.drop(columns = ['products', 'activityControls', 'details', 'source', 'subtitles']
#Filtering out the channel names from the subtitles column
df['channelName'] = df['subtitles'].apply(lambda x: x[0]['name'] if isinstance(x, list) and len(x)> 0 else None)
df
```

```python
#Checking if there are any other prefixes other than watched in the title column by getting the first word in each entry
df['title'].str.split(" ", n = 1).str[0].value_counts()
```

```python
#Filtering out any records that were not watched videos

df = df[df['title'].str.startswith('Watched ')]
df
```

```python
#Removing the Watched prefix to have only video titles in the column
df['title'] = df['title'].str.replace('^Watched ', '', regex = True)
df
```

```python
#Converting time format to something human legible
df['time'] = pd.to_datetime(df['time'], errors = 'coerce')
df
```

```python
#Dropping unnecessary columns
df = df.drop(columns = ['header', 'subtitles', 'products', 'activityControls', 'description', 'details', 'source'])
df
```

```python
#Resetting the index counts
df.reset_index(drop = True)
```

```python
#Importing my subscription data to join it with my watch history
df2 = pd.read_csv(r"C:\Users\0106o\OneDrive\Pictures\Documents\Datasets\Youtube Project Files\subscriptions\subscriptions.csv")
df2
```

```python
#Creating a true or false column to check if the watched videos are from one of my subscriptions
df['subscribed'] = df['channelName'].isin(df2['Channel Title'])
df
```

```python
#Checking if the ratio of subscribed content to unsubscribed content makes sense
df['subscribed'].value_counts()
```
